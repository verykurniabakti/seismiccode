```python
import json
import os
import numpy as np
import matplotlib.pyplot as plt

def inspect_stead_json_robust(json_file_path):
    print("=" * 80)
    print(f"MENGECEK STRUKTUR FILE JSON: {os.path.basename(json_file_path)}")
    print("=" * 80)
    
    if not os.path.exists(json_file_path):
        print(f"[ERROR] File tidak ditemukan di path: {json_file_path}")
        return

    print("Membaca file JSON (memuat data ke memori)...")
    try:
        with open(json_file_path, 'r') as f:
            raw_data = json.load(f)
    except Exception as e:
        print(f"[ERROR] Gagal memparsing JSON: {e}")
        return
        
    # --- PENANGANAN FLEKSIBEL STRUKTUR JSON ---
    data_list = []
    
    if isinstance(raw_data, list):
        print("[INFO] Format terdeteksi: List of Dictionaries (Standar)")
        data_list = raw_data
        
    elif isinstance(raw_data, dict):
        print("[INFO] Format terdeteksi: Dictionary Tunggal / Nested Keys")
        print(f"Key utama pada root level: {list(raw_data.keys())}")
        
        # Cek apakah key tertentu langsung berisi list
        possible_keys = ['data', 'traces', 'signals', 'records']
        found_key = next((k for k in possible_keys if k in raw_data), None)
        
        if found_key and isinstance(raw_data[found_key], list):
            data_list = raw_data[found_key]
            print(f"[INFO] Mengambil data dari key utama: '{found_key}'")
        else:
            print("[INFO] Mengubah dictionary key-value menjadi list of records...")
            for k, v in raw_data.items():
                if isinstance(v, dict):
                    v['trace_key_id'] = k  # simpan ID aslinya
                    data_list.append(v)
                    
    if len(data_list) == 0:
        print("[ERROR] Tidak dapat mengekstrak data list dari struktur JSON ini.")
        print("Struktur root data:", type(raw_data))
        return

    print(f"Total record/sampel yang berhasil dimuat: {len(data_list)}")
    print("=" * 80)
    
    # Ambil dan tampilkan ringkasan 10 sampel pertama
    samples_to_show = data_list[:10]
    for idx, sample in enumerate(samples_to_show, 1):
        print(f"\n--- Sampel #{idx} ---")
        if isinstance(sample, dict):
            t_name = sample.get('trace_name', sample.get('trace_key_id', 'N/A'))
            station = sample.get('receiver_code', sample.get('station', 'N/A'))
            mag = sample.get('source_magnitude', sample.get('magnitude', 'N/A'))
            
            print(f"  • Trace / ID   : {t_name}")
            print(f"  • Stasiun      : {station}")
            print(f"  • Magnitudo    : {mag}")
            
            # Cek komponen gelombang
            for comp in ['trace_z', 'trace_n', 'trace_e', 'z', 'n', 'e']:
                if comp in sample and isinstance(sample[comp], (list, np.ndarray)):
                    print(f"  • Komponen [{comp}] : {len(sample[comp])} titik sampel")
        else:
            print(f"  • Format sampel tidak berupa dictionary: {type(sample)}")

if __name__ == "__main__":
    target_json_path = "/Volumes/Local Disk/Code_Git/S3_code/seismic/mcu_quake/code_gen_trying/indonesia_jaya_benchmarking_data/mcquake_ori_file/Benchmark_ STEAD 3C_ test n15275 r100/STEAD data, test n15275 r100.json"
    inspect_stead_json_robust(target_json_path)
```

```python
import json
import os
import numpy as np

def inspect_nested_stead_json(json_file_path):
    print("=" * 80)
    print(f"MEMBEDAH ISI FILE: {os.path.basename(json_file_path)}")
    print("=" * 80)
    
    with open(json_file_path, 'r') as f:
        raw_data = json.load(f)
        
    if not isinstance(raw_data, dict):
        print("[INFO] Data bukan dictionary bertingkat.")
        return
        
    # Ambil ID trace pertama sebagai sampel
    first_trace_id = list(raw_data.keys())[0]
    first_sample_dict = raw_data[first_trace_id]
    
    print(f"Total total trace dalam file: {len(raw_data.keys())}")
    print(f"Contoh Trace ID Pertama: {first_trace_id}\n")
    
    print("=" * 50)
    print("DAFTAR KOLOM/KEY YANG TERSEDIA PADA SAMPEL PERTAMA:")
    print("=" * 50)
    for key, val in first_sample_dict.items():
        if isinstance(val, list):
            print(f" • {key} : Berbentuk List/Array dengan panjang {len(val)}")
        else:
            print(f" • {key} : {val}")

if __name__ == "__main__":
    target_json_path = "/Volumes/Local Disk/Code_Git/S3_code/seismic/mcu_quake/code_gen_trying/indonesia_jaya_benchmarking_data/mcquake_ori_file/Benchmark_ STEAD 3C_ test n15275 r100/STEAD data, test n15275 r100.json"
    inspect_nested_stead_json(target_json_path)
```

```python
import os
import pandas as pd

def inspect_stead_csv_metadata(csv_file_path):
    print("=" * 80)
    print(f"MEMBEDAH FILE CSV METADATA: {os.path.basename(csv_file_path)}")
    print("=" * 80)
    
    if not os.path.exists(csv_file_path):
        print(f"[ERROR] File CSV tidak ditemukan pada path: {csv_file_path}")
        return

    print("Memuat file CSV ke Pandas DataFrame...")
    try:
        # Membaca file CSV
        df = pd.read_csv(csv_file_path)
    except Exception as e:
        print(f"[ERROR] Gagal membaca file CSV: {e}")
        return
        
    # 1. Informasi Umum Dataset
    print(f"\n[INFO UMUM]")
    print(f" • Total Baris (Sampel) : {df.shape[0]}")
    print(f" • Total Kolom (Atribut): {df.shape[1]}")
    
    # 2. Daftar Nama Kolom
    print(f"\n[DAFTAR KOLOM / FITUR DALAM CSV]")
    for idx, col in enumerate(df.columns, 1):
        non_null_count = df[col].notnull().sum()
        dtype = df[col].dtype
        print(f" {idx:2d}. {col:<25} | Tipe: {str(dtype):<10} | Terisi: {non_null_count} baris")
        
    # 3. Menampilkan 10 Sampel Pertama
    print(f"\n[10 SAMPEL PERTAMA DARI DATASET]")
    print("-" * 80)
    # Menampilkan tabel ringkas 10 baris pertama
    print(df.head(10).to_string())
    
    # 4. Ringkasan Statistik untuk kolom SNR (jika ada)
    snr_columns = [c for c in df.columns if 'snr' in c.lower() or 'db' in c.lower()]
    if snr_columns:
        print(f"\n[RINGKASAN STATISTIK KOLOM SNR]")
        print(df[snr_columns].describe().to_string())

if __name__ == "__main__":
    # Sesuaikan path lokal Anda di sini
    target_csv_path = "/Volumes/Local Disk/Code_Git/S3_code/seismic/mcu_quake/code_gen_trying/indonesia_jaya_benchmarking_data/mcquake_ori_file/Benchmark_ STEAD 3C_ test n15275 r100/STEAD 3C meta, test SNR_DB, n15275.csv"
    
    inspect_stead_csv_metadata(target_csv_path)
```
